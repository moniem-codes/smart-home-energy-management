```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy
import skfuzzy as fuzz

print("Libraries are ready")
```

```python
# Smart Home Energy Management System

# Inputs
# 1. Occupancy: Number of people
# 2. Temperature: Room temperature
# 3. Natural Light: Light intensity

# Outputs
# 1. AC Power: Air conditioner power
# 2. Lighting Power: Lighting power

print("Smart Home Fuzzy Logic Project")
print("Inputs and outputs are defined.")
```

```python
import numpy as np

occupancy_range = np.arange(0, 11, 1)
temperature_range = np.arange(15, 36, 1)
light_range = np.arange(0, 101, 1)

power_range = np.arange(0, 101, 1)

print("Occupancy:", occupancy_range)
print("Temperature:", temperature_range)
print("Natural Light:", light_range)
print("Power:", power_range)
```

```python
import skfuzzy as fuzz

# Occupancy
occupancy_zero = fuzz.trapmf(occupancy_range, [0, 0, 0, 1])
occupancy_few = fuzz.trapmf(occupancy_range, [1, 2, 3, 5])
occupancy_many = fuzz.trapmf(occupancy_range, [4, 6, 10, 10])

# Temperature
temperature_cold = fuzz.trapmf(temperature_range, [15, 15, 20, 22])
temperature_normal = fuzz.trapmf(temperature_range, [21, 24, 25, 27])
temperature_hot = fuzz.trapmf(temperature_range, [26, 30, 35, 35])

# Natural Light
light_dark = fuzz.trapmf(light_range, [0, 0, 20, 40])
light_dim = fuzz.trapmf(light_range, [30, 50, 50, 70])
light_bright = fuzz.trapmf(light_range, [60, 80, 100, 100])

print("Membership functions created successfully.")
```

```python
# Occupancy
occupancy_zero = fuzz.trapmf(occupancy_range, [0, 0, 0, 1])
occupancy_few = fuzz.trapmf(occupancy_range, [0, 1, 3, 5])
occupancy_many = fuzz.trapmf(occupancy_range, [3, 5, 10, 10])

# Temperature
temperature_cold = fuzz.trapmf(temperature_range, [15, 15, 19, 23])
temperature_normal = fuzz.trapmf(temperature_range, [20, 23, 26, 29])
temperature_hot = fuzz.trapmf(temperature_range, [26, 30, 35, 35])

# Natural Light
light_dark = fuzz.trapmf(light_range, [0, 0, 20, 45])
light_dim = fuzz.trapmf(light_range, [30, 45, 55, 75])
light_bright = fuzz.trapmf(light_range, [60, 80, 100, 100])

print("Membership functions updated successfully.")
```

```python
rules = []

for occupancy_label in ["Zero", "Few", "Many"]:
    for temperature_label in ["Cold", "Normal", "Hot"]:
        for light_label in ["Dark", "Dim", "Bright"]:

            if occupancy_label == "Zero":
                ac_power = 0
                lighting_power = 0

            else:
                if temperature_label == "Cold":
                    ac_power = 0
                elif temperature_label == "Normal":
                    ac_power = 0 if occupancy_label == "Few" else 50
                else:
                    ac_power = 50 if occupancy_label == "Few" else 100

                if light_label == "Dark":
                    lighting_power = 100
                elif light_label == "Dim":
                    lighting_power = 50
                else:
                    lighting_power = 0

            rules.append({
                "Occupancy": occupancy_label,
                "Temperature": temperature_label,
                "Natural Light": light_label,
                "AC Power": ac_power,
                "Lighting Power": lighting_power
            })

rules_df = pd.DataFrame(rules)

print("Number of rules:", len(rules_df))
display(rules_df)
```

```python
def calculate_fuzzy_output(occupancy, temperature, natural_light):

    # Calculate occupancy membership degrees
    occupancy_degrees = {
        "Zero": fuzz.interp_membership(
            occupancy_range, occupancy_zero, occupancy
        ),
        "Few": fuzz.interp_membership(
            occupancy_range, occupancy_few, occupancy
        ),
        "Many": fuzz.interp_membership(
            occupancy_range, occupancy_many, occupancy
        )
    }

    # Calculate temperature membership degrees
    temperature_degrees = {
        "Cold": fuzz.interp_membership(
            temperature_range, temperature_cold, temperature
        ),
        "Normal": fuzz.interp_membership(
            temperature_range, temperature_normal, temperature
        ),
        "Hot": fuzz.interp_membership(
            temperature_range, temperature_hot, temperature
        )
    }

    # Calculate natural light membership degrees
    light_degrees = {
        "Dark": fuzz.interp_membership(
            light_range, light_dark, natural_light
        ),
        "Dim": fuzz.interp_membership(
            light_range, light_dim, natural_light
        ),
        "Bright": fuzz.interp_membership(
            light_range, light_bright, natural_light
        )
    }

    # Calculate rule firing strengths
    weighted_ac = 0
    weighted_light = 0
    total_strength = 0

    for rule in rules:

        alpha = min(
            occupancy_degrees[rule["Occupancy"]],
            temperature_degrees[rule["Temperature"]],
            light_degrees[rule["Natural Light"]]
        )

        if alpha > 0:
            weighted_ac += alpha * rule["AC Power"]
            weighted_light += alpha * rule["Lighting Power"]
            total_strength += alpha

    # Calculate Sugeno outputs
    if total_strength == 0:
        ac_power = 0
        lighting_power = 0
    else:
        ac_power = weighted_ac / total_strength
        lighting_power = weighted_light / total_strength

    return ac_power, lighting_power


print("Fuzzy controller function created successfully!")
```

```python
occupancy = 2
temperature = 30
natural_light = 40

ac_power, lighting_power = calculate_fuzzy_output(
    occupancy,
    temperature,
    natural_light
)

print("Smart Home Control Results")
print("---------------------------")
print("Occupancy:", occupancy, "people")
print("Temperature:", temperature, "C")
print("Natural Light:", natural_light, "%")
print("AC Power:", round(ac_power, 2), "%")
print("Lighting Power:", round(lighting_power, 2), "%")
```

```python
test_cases = [
    (0, 25, 50),
    (2, 30, 40),
    (5, 32, 20),
    (3, 22, 80),
    (8, 28, 10)
]

results = []

for occupancy, temperature, natural_light in test_cases:

    ac_power, lighting_power = calculate_fuzzy_output(
        occupancy,
        temperature,
        natural_light
    )

    results.append({
        "Occupancy": occupancy,
        "Temperature": temperature,
        "Natural Light": natural_light,
        "AC Power": round(ac_power, 2),
        "Lighting Power": round(lighting_power, 2)
    })

results_df = pd.DataFrame(results)

display(results_df)
```

```python
# Cell 11: Visualize Fuzzy Controller Results

plt.figure(figsize=(10, 5))

plt.plot(
    range(len(results_df)),
    results_df["AC Power"],
    marker="o",
    label="AC Power"
)

plt.plot(
    range(len(results_df)),
    results_df["Lighting Power"],
    marker="s",
    label="Lighting Power"
)

plt.xlabel("Test Case")
plt.ylabel("Power Level (%)")
plt.title("Smart Home Energy Management Results")
plt.xticks(range(len(results_df)), [f"Case {i+1}" for i in range(len(results_df))])
plt.ylim(0, 100)
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Cell 12: Validate the Fuzzy Controller

print("Testing Fuzzy Controller")
print("-------------------------")

# Test 1: No people in the house
ac, light = calculate_fuzzy_output(0, 25, 50)
print("Test 1 - No People:")
print("AC Power:", round(ac, 2), "%")
print("Lighting Power:", round(light, 2), "%")

# Test 2: High temperature and low natural light
ac, light = calculate_fuzzy_output(5, 32, 20)
print("\nTest 2 - Hot Temperature and Low Light:")
print("AC Power:", round(ac, 2), "%")
print("Lighting Power:", round(light, 2), "%")

# Test 3: Normal temperature and bright natural light
ac, light = calculate_fuzzy_output(3, 25, 90)
print("\nTest 3 - Normal Temperature and Bright Light:")
print("AC Power:", round(ac, 2), "%")
print("Lighting Power:", round(light, 2), "%")
```

```python
# Cell 13: Effect of Occupancy on Power Consumption

occupancy_values = range(0, 11)

ac_values = []
lighting_values = []

for people in occupancy_values:
    ac, light = calculate_fuzzy_output(
        people,
        30,
        20
    )

    ac_values.append(ac)
    lighting_values.append(light)

plt.figure(figsize=(10, 5))

plt.plot(
    list(occupancy_values),
    ac_values,
    marker="o",
    label="AC Power"
)

plt.plot(
    list(occupancy_values),
    lighting_values,
    marker="s",
    label="Lighting Power"
)

plt.xlabel("Number of People")
plt.ylabel("Power Level (%)")
plt.title("Effect of Occupancy on AC and Lighting")
plt.xticks(range(0, 11))
plt.ylim(0, 100)
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Cell 14: Compare Energy Management Results

comparison_df = results_df.copy()

comparison_df["Total Power Level"] = (
    comparison_df["AC Power"] +
    comparison_df["Lighting Power"]
)

display(comparison_df)

print("Average AC Power:",
      round(comparison_df["AC Power"].mean(), 2), "%")

print("Average Lighting Power:",
      round(comparison_df["Lighting Power"].mean(), 2), "%")

print("Average Total Power Level:",
      round(comparison_df["Total Power Level"].mean(), 2), "%")
```
